# Data Preprocessing

### 1.Handling Missing Vlaues

```python
df['Income'] = df['Income'].fillna(df['Income'].median())
```

# Feature Engineering

```python
#Age
df['Age'] = 2026 - df['Year_Birth']
```

```python
#Customer Joining Date
df['Dt_Customer'] = pd.to_datetime(df["Dt_Customer"],dayfirst = True)

reference_date = df['Dt_Customer'].max()

df['Customer_Tenure'] = (reference_date - df["Dt_Customer"]).dt.days

df.head(1)
```

```python
df.columns
```

```python
# Total Spending

df['Total_spending'] = df['MntWines'] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
# Total Purchases

df["Total_Purchases"] = df["NumDealsPurchases"] + df["NumWebPurchases"] + df["NumCatalogPurchases"] + df["NumStorePurchases"]
```

```python
# Total Children

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
# Purchase to income ration

df["Purchase to income"] = df['Total_Purchases'] / df['Income']
```

```python
df['Education'] = df['Education'].replace({
    'Basic': 'Undergraduate',
    '2n Cycle': 'Undergraduate',
    'Graduation': "Graduate",
    'Master': 'PostGraduate',
    'PhD': 'PostGraduate'
})
```

```python
df['Living_With'] = df['Marital_Status'].replace({
    'Married':'Partner',
    'Together':'Partner',
    'Single': 'Alone',
    'Divorced': 'Alone',
    "Absurd":  'Alone',
    "YOLO":  'Alone'
})
```

```python
df.columns
```

### Scaling and Visualization

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[scale_cols])

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

df['PC1'] = X_pca[:, 0]
df['PC2'] = X_pca[:, 1]
df['PC3'] = X_pca[:, 2]  

df.drop(columns=scale_cols, inplace=True)
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    X_pca[:, 2]
)

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")

plt.show()
```

```python
pca.explained_variance_ratio_
```

### Analyze k value

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters = k, random_state = 42)
    kmeans.fit_predict(df)
    wcss.append(kmeans.inertia_)
    
```

```python
knee = KneeLocator(range(1,11),wcss, curve = "convex", direction = "decreasing")
optimal_k = knee.elbow
print(optimal_k)

plt.plot(range(1,11), wcss, marker = "o")
plt.xlabel("k")
plt.ylabel("wcss")
```

```python
##Silhouette Score
from sklearn.metrics import silhouette_score

scores = []

for k in range(2,11):
    kmeans = KMeans(n_clusters = k, random_state = 42)
    labels = kmeans.fit_predict(df)
    score = silhouette_score(df,labels)
    scores.append(score)

plt.plot(range(2,11), scores, marker = "o")
plt.xlabel("k")
plt.ylabel("scores")
```

```python
k_range = range(2, 11)

fig, axis = plt.subplots(figsize=(8, 6))

axis.plot(k_range, wcss[1:], marker="o")
axis.set_xlabel("K")
axis.set_ylabel("WCSS")

axs2 = axis.twinx()

axs2.plot(k_range, scores, marker="x", linestyle="--")
axs2.set_ylabel("Silhouette Score")

plt.show()
```
